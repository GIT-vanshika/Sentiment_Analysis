```python
hash_counts = df1["Transaction Hash"].value_counts()
repeated_hashes = hash_counts[hash_counts > 1].index
repeated_hashes[:5]
```

```python
sample_hash=repeated_hashes[0]
sample_rows=df1[df1["Transaction Hash"]== sample_hash]
print(sample_rows[[
    "Account", "Coin", "Timestamp", 
    "Size USD", "Closed PnL", "Fee"]])
```

```python
daily_df= (
  df1.groupby(["Account", "trade_date"]).agg(
    total_daily_pnl= ("Closed PnL", "sum"),
    num_trades = ("Closed PnL", "count"),
    wins= ("is_win", "sum"),
    avg_size_used = ("Size USD", "mean"),
    avg_fee= ("Fee", "mean")
  ).reset_index()
)

daily_df["win_rate"]= daily_df["wins"] /daily_df["num_trades"]
```
